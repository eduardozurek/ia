**Ejemplo - Categoricas Binarizadas - KNN**


```
import pandas as pd
#url = "https://web.stanford.edu/~hastie/ElemStatLearn/datasets/marketing.data"
datos = pd.read_csv("marketing.data",sep=' ',header=None)
```

```
datos
```

```
import numpy as np
datos = np.asarray(datos)
print(datos.shape)
print(datos)
```

Como se puede ver, este conjunto de datos contiene varios "nan".

Lo primero que debemos hacer es remmover esos "nan" para evitar errores.

```
contar_nan=np.isnan(datos)
filas_con_nan = contar_nan.sum(axis=1)
print(filas_con_nan.shape)
print(np.sum(filas_con_nan>0))
print(np.sum(filas_con_nan==0))
```

Esto me dice que hay, 2117 filas que contienen "nan"\
y 6876 fila que no contienen "nan".

Los metodos de clasificación estudiados aquí\
no tienen manera de lidiar con los "nan",\
entonces trabajaremos solo con las filas que\
no tienen "nan"

```
datos = datos[filas_con_nan==0,:]
# y volvemos a revisar
contar_nan=np.isnan(datos)
filas_con_nan = contar_nan.sum(axis=1)
print(filas_con_nan.shape)
print(np.sum(filas_con_nan>0))
print(np.sum(filas_con_nan==0))
```

```
datos
```

Ahora sí ...\
Procedemos binarizar las categorías de X y a entrenar y probar el clasificador

```
from sklearn import preprocessing
lb = preprocessing.LabelBinarizer()
y = datos[:,0]
print(y)
X = datos[:,1:15]
print(X.shape)
xt = X[:,0]
lb.fit(xt)
X_binarizada = lb.transform(xt)
for i in range(1,13):
  xt = X[:,i]
  lb.fit(xt)
  X_binarizada = np.append(X_binarizada,lb.transform(xt),axis=1)
print(X_binarizada.shape)
X_binarizada = X_binarizada>0
X=X_binarizada
```

```
X_binarizada
```

```
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = \
    train_test_split(X, y, test_size=0.33, random_state=17)
```

```
from sklearn import neighbors
clfKNN = neighbors.KNeighborsClassifier(n_neighbors=101, # número de vecinos
                                        weights='uniform', # pesos asignados a los vecinos al momento
                                                           # de tomar la decisión
                                                           # 'uniform', 'distance'
                                        algorithm='brute', # optimiza la ejecución del clasificador
                                                           # ‘auto’, ‘ball_tree’, ‘kd_tree’, ‘brute’
                                        metric='jaccard') #jaccard
y_esperado = y_test
clfKNN.fit(X_train,y_train)
y_predicho = clfKNN.predict(X_test)
print(X_train)
```

```
from sklearn.metrics import accuracy_score
print(accuracy_score(y_esperado,y_predicho))
```

Como se puede ver, el sistema es capaz de clasificar\
correctamente menos del 31% de los datos de prueba

Pero que pasa si probamos el sistema solo con\
las dos categorías más bajas y las dos más altas

Veamos ...

```
# con la siguiente línea vemos las diferentes categorías
print(np.unique(y))
indice = (y==1.0)|(y==2.0)|(y==8.0)|(y==9.0)
y1 = y[indice]
X1 = X[indice,:]
print(X1.shape)
print(y1.shape)
X_train, X_test, y_train, y_test = \
    train_test_split(X1, y1, test_size=0.33, random_state=17)
clfKNN.fit(X_train,y_train)
y_predicho = clfKNN.predict(X_test)
y_esperado = y_test
print(accuracy_score(y_esperado,y_predicho))
```

El clasificador mejoró!
