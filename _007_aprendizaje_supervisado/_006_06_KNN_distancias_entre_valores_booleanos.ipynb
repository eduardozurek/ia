# Ejemplo KNN con distacias basadas en valores booleanos

En algunos conjuntos de datos predominan los valores booleanos, es decir, presencia o ausencia de una característica.

En este tipo de conjuntos de datos se aplican otras fórmulas de distancia: Jaccard, Dice, etc.

Una descripción de estas medidas de distancia se encuentra en:

https://scikit-learn.org/0.24/modules/generated/sklearn.neighbors.DistanceMetric.html#

Para este ejemplo usaremos los datos Spam.

```
import pandas as pd
url = "https://web.stanford.edu/~hastie/ElemStatLearn/datasets/spam.data"
datos = pd.read_csv(url,sep=' ',header=None)
```

Para aplicar KNN sobre los datos Spam, debemos binarizarlos.

```
import numpy as np
datos = np.asarray(datos)
y = datos[:,-1]
X = datos[:,0:48]
print(y.shape)
print(X.shape)
X = X>0.1
```

```
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = \
    train_test_split(X, y, test_size=0.25, random_state=23)
```


Para este ejemplo aplicaremos Jaccard.

```
from sklearn import neighbors
clfKNN = neighbors.KNeighborsClassifier(n_neighbors=55, # número de vecinos
                                        weights='distance', # pesos asignados a los vecinos al momento
                                                           # de tomar la decisión
                                                           # 'uniform', 'distance'
                                        algorithm='brute', # optimiza la ejecución del clasificador
                                                           # ‘auto’, ‘ball_tree’, ‘kd_tree’, ‘brute’
                                        #metric='jaccard'
                                        #metric='matching'
                                        #metric='dice'
                                        #metric='kulsinski'
                                        #metric='rogerstanimoto'
                                        #metric='russellrao'
                                        #metric='sokalmichener'
                                        metric='sokalsneath'
                                        )
clfKNN.fit(X_train,y_train)
y_esperado = y_test
y_predicho = clfKNN.predict(X_test)
```

```
from sklearn.metrics import accuracy_score, confusion_matrix, recall_score, precision_score
print(accuracy_score(y_esperado,y_predicho))
print(recall_score(y_esperado,y_predicho))
print(precision_score(y_esperado,y_predicho))
print(confusion_matrix(y_esperado,y_predicho))
```
