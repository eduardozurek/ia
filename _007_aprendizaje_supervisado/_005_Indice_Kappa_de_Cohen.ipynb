El **índice Kappa de Cohen (κ)** es una métrica **útil y complementaria** para evaluar algoritmos de clasificación supervisada, especialmente cuando interesa medir el acuerdo entre las clases reales y las predichas **corrigiendo el acuerdo que podría ocurrir por azar**. Sin embargo, no suele ser suficiente como métrica única.

Su definición básica es:

$$
\kappa=\frac{p_o-p_e}{1-p_e}
$$

donde \($p_o$\) es el **acuerdo observado** —equivalente a la *accuracy*— y \($p_e$\) es el **acuerdo esperado por azar**, calculado a partir de las distribuciones marginales de clases reales y predichas.

### ¿Qué aporta frente a Accuracy?

Suponga un problema binario con fuerte desbalance:

* Clase 0: 950 muestras
* Clase 1: 50 muestras

Un clasificador que **siempre prediga clase 0** tendría:

$$
Accuracy=\frac{950}{1000}=0.95
$$

A primera vista, 95 % parece excelente. Pero el clasificador realmente no ha aprendido a detectar la clase 1.

Para ese clasificador, el acuerdo esperado por azar según los marginales también es:

$$
p_e=(0.95)(1.0)+(0.05)(0)=0.95
$$

por lo tanto:

$$
\kappa=\frac{0.95-0.95}{1-0.95}=0
$$

Esto muestra una propiedad interesante: **95 % de accuracy, pero κ = 0**. Kappa revela que el desempeño no supera lo esperado dadas esas distribuciones marginales.

### Interpretación

En términos generales:

| \($\kappa$) | Interpretación conceptual                |
| ---------: | ---------------------------------------- |
|      \(1\) | Acuerdo perfecto                         |
|      \(0\) | Acuerdo equivalente al esperado por azar |
|     \(<0\) | Acuerdo peor que el esperado por azar    |

Existen escalas que asignan etiquetas como *moderado*, *sustancial*, etc., a ciertos intervalos de κ. Conviene utilizarlas con cautela: los puntos de corte son convenciones y dependen del dominio.

### ¿Qué tan relevante es para Machine Learning?

Es una **métrica secundaria bastante valiosa**, no como reemplazo de las métricas principales. Su relevancia aumenta cuando hay **desbalance de clases**, cuando se comparan clasificadores cuyas distribuciones de predicción son diferentes o cuando el concepto de *agreement* tiene importancia en el problema.

Para una clase de **evaluación de algoritmos de aprendizaje supervisado**, tiene sentido presentarlo después de la matriz de confusión y de *accuracy*, porque Kappa se deriva naturalmente de ellas:

$$
\boxed{\text{Matriz de confusión}
\rightarrow Accuracy
\rightarrow \text{Acuerdo esperado}
\rightarrow Cohen's Kappa}
$$

Después se contrastaría con métricas que responden preguntas diferentes:

$$
\text{Precision},\quad
\text{Recall/Sensitivity},\quad
\text{Specificity},\quad
F_1,\quad
\text{ROC-AUC}\quad
$$

La diferencia pedagógica importante es que **Kappa no mide exactamente lo mismo que ROC-AUC o F1**. Kappa evalúa el acuerdo global corregido por azar a partir de las predicciones finales; ROC-AUC evalúa la capacidad de ordenamiento/discriminación a través de múltiples umbrales; F1 concentra la evaluación en precision y recall de la clase positiva.

Un punto adicional importante: **Kappa también tiene limitaciones**. Puede verse afectado por la prevalencia de las clases y por los marginales de las predicciones. Por eso, en clasificación desbalanceada no diría simplemente que "Kappa soluciona el problema de accuracy"; más precisamente, **aporta información adicional que accuracy oculta**.

Para el tipo de ejemplos que está preparando con **Random Forest, matriz de confusión y curva ROC**, una secuencia didáctica particularmente clara sería:

$$
\boxed{
\text{Confusion Matrix}
\rightarrow
\text{Accuracy}
\rightarrow
\text{Precision/Recall}
\rightarrow
F_1
\rightarrow
\kappa
\rightarrow
\text{ROC}
\rightarrow
\text{AUC}
}
$$

y se puede usar **el mismo conjunto de \(y_{\text{true}}\) y \(y_{\text{pred}}\)** para calcular manualmente Kappa paso a paso antes de verificarlo con `sklearn.metrics.cohen_kappa_score`. Esto permite que los estudiantes entiendan de dónde sale κ en lugar de verlo simplemente como otra función de sklearn.


En el **índice Kappa de Cohen**, \($P_e$\) representa la **proporción de acuerdos esperados por azar**, suponiendo que tanto las clases reales como las predichas mantienen las proporciones indicadas por sus marginales.

Supongamos una matriz de confusión binaria:

|                    | Predicho \(+\) | Predicho \(-\) | Total real |
| ------------------ | -------------: | -------------: | ---------: |
| **Real \(+\)**     |             TP |             FN |  \(TP+FN\) |
| **Real \(-\)**     |             FP |             TN |  \(FP+TN\) |
| **Total predicho** |      \(TP+FP\) |      \(FN+TN\) |      \(N\) |

Primero calculamos las proporciones marginales:

$$
P(\text{Real }+) = \frac{TP+FN}{N}
$$

$$
P(\text{Predicho }+) = \frac{TP+FP}{N}
$$

y, para la clase negativa:

$$
P(\text{Real }-) = \frac{FP+TN}{N}
$$

$$
P(\text{Predicho }-) = \frac{FN+TN}{N}
$$

Entonces, el acuerdo esperado por azar es:

$$
\boxed{
P_e =
P(\text{Real }+)P(\text{Predicho }+)
+
P(\text{Real }-)P(\text{Predicho }-)
}
$$

Es decir:

$$
\boxed{
P_e =
\frac{TP+FN}{N}\frac{TP+FP}{N}
+
\frac{FP+TN}{N}\frac{FN+TN}{N}
}
$$

### Ejemplo

Consideremos:

|                | Predicho \(+\) | Predicho \(-\) |   Total |
| -------------- | -------------: | -------------: | ------: |
| **Real \(+\)** |             40 |             10 |  **50** |
| **Real \(-\)** |             20 |             30 |  **50** |
| **Total**      |         **60** |         **40** | **100** |

Los marginales expresados como proporciones son:

$$
P(\text{Real }+) = \frac{50}{100}=0.50
$$

$$
P(\text{Predicho }+) = \frac{60}{100}=0.60
$$

$$
P(\text{Real }-) = \frac{50}{100}=0.50
$$

$$
P(\text{Predicho }-) = \frac{40}{100}=0.40
$$

Por tanto:

$$
P_e=(0.50)(0.60)+(0.50)(0.40)
$$

$$
P_e=0.30+0.20
$$

$$
\boxed{P_e=0.50}
$$

La idea es que existen **dos maneras de coincidir por azar**: ambos asignan la clase positiva, o ambos asignan la clase negativa. Por eso se multiplican los marginales correspondientes y luego se suman:

$$
\boxed{
P_e=\sum_{i=1}^{K}P(\text{Real}=i)\,P(\text{Predicho}=i)
}
$$

Esta última expresión es la generalización para un problema de **\(K\) clases**.
