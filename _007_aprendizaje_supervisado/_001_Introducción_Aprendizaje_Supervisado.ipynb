El aprendizaje supervisado es un paradigma de la inteligencia artificial y el aprendizaje automático (machine learning) en el que un modelo se entrena utilizando un conjunto de datos que incluye tanto las entradas (features) como las salidas deseadas (etiquetas o labels).

El objetivo es que el modelo aprenda una función que relacione las entradas con las salidas, de modo que pueda predecir correctamente la salida para nuevos datos que no ha visto antes.

## Características principales

Datos etiquetados: cada ejemplo de entrenamiento incluye la variable de entrada y la respuesta correcta.

Función objetivo: consiste en aprender una función $f: X \to Y$ que aproxime de la mejor manera posible la relación entre las variables de entrada $X$ y las salidas $Y$.

Evaluación: se mide con métricas como exactitud, error cuadrático medio, precisión, recall, etc.

## Tipos de problemas

Clasificación: la salida es una categoría discreta.
Ejemplo: identificar si un correo es spam o no spam.

Regresión: la salida es un valor numérico continuo.
Ejemplo: predecir el precio de una casa según sus características.

## Ejemplos de algoritmos

- Clasificación:

  - Regresión logística

  - Máquinas de soporte vectorial (SVM)

  - Árboles de decisión y Random Forest

  - Redes neuronales

- Regresión:

  - Regresión lineal

  - Regresión polinómica

  - Redes neuronales para regresión

  - Gradient Boosting Regressor

## Flujo típico del aprendizaje supervisado

1. Recolección de datos → conjunto de entrenamiento y prueba.

2. Preprocesamiento → limpieza, normalización, codificación.

3. Entrenamiento → ajuste de parámetros del modelo.

4. Validación y prueba → medir desempeño en datos no vistos.

5. Implementación → usar el modelo en producción.

## Taxonomía de técnicas de Aprendizaje Supervisado

### 1. Modelos Lineales

- Regresión lineal (para variables continuas).

- Regresión logística (para clasificación binaria/multiclase).

- Modelos lineales regularizados: Ridge, Lasso, Elastic Net.

- Análisis Discriminante:

  - LDA (Linear Discriminant Analysis).

  - QDA (Quadratic Discriminant Analysis).

### 2. Modelos Basados en Árboles

- Árboles de decisión.

- Conjuntos de árboles (Ensemble Methods):

  - Random Forest.

  - Gradient Boosting Machines (GBM).

  - XGBoost, LightGBM, CatBoost.

### 3. Métodos de Vecindad

- K-Nearest Neighbors (KNN).

- R-Nearest Neighbors (KNN).

### 4. Máquinas de Soporte Vectorial (SVM)

- Clasificación con kernel lineal.

- Clasificación con kernels no lineales (polinómico, RBF, sigmoide).

- SVR (Support Vector Regression).

### 5. Modelos Probabilísticos

- Naïve Bayes (Bernoulli, Multinomial, Gaussiano).

- Modelos gráficos probabilísticos supervisados.

### 6. Redes Neuronales y Deep Learning

- Perceptrón multicapa (MLP).

- Redes convolucionales (CNN) → clasificación de imágenes.

- Redes recurrentes (RNN, LSTM, GRU) → series temporales y texto.

- Transformers supervisados → NLP, visión.

### 7. Métodos de Ensamble

(algunos ya listados en árboles, pero aquí como categoría general)

- Bagging (Bootstrap Aggregating).

- Boosting (AdaBoost, Gradient Boosting, XGBoost, etc.).

- Stacking (combinación de varios modelos base).
