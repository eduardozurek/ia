**Clasificador Bayesiano Ingenuo - Naïve Bayes Classifier**

Partiendo del Teorema de Bayes:

$P(y/X)=\frac{P(X/y)P(y)}{P(X)}$,_ecuación 1_

Donde:
* $P(y/X)$ es la probabilidad condicional de $y$ dado $X$
* $P(X/y)$ es la probabilidad condicional de $X$ dado $y$
* $P(X)$ es la probabilidad marginal de $X$
* $P(y)$ es la probabilidad marginal de $Y$

Si $y$ es la salida del sistema, $X=(x_0,x_1,...,x_n)$ un conjunto de característica de $y$\
y nuestro sistema es binario (es decir $y$ es verdadero o falso ), entonces:

Se estima el valor de $y$ dado un conjunto de características $X$ calculando las probabilidades:

* $P(y=0/X)$
* $P(y=1/X)$

Y se asigna a $y$ el valor que tenga la mayor probabilidad.


Si usamos la _ecuación_ 1 para calcular las probabilidades,se hace evidente que\
el denominador $P(X)$ no afecta en la toma de la decisión porque es igual en ambos casos.

Aplicando la regla de la cadena, el numerador de la _ecuación 1_ es equivalente a:

$P(X,y)P(y)=P(y,X)=P(y,x_0,x_1,...,x_n)$

Además:

$P(y,x_0.x_1,...,x_n)=P(x_0,x_1,...,x_n,y)$

Aplicando nuevamente la regla de la cadena:

$P(x_0.x_1,...,x_n,y)$

$=P(x_0/x_1,...,x_n,y)P(x_1/x_2,...,x_n,y)...P(x_n/y)P(y)$

Es en este punto donde se aplica la ingenuidad de la estimación, ya que\
se supone que las características son mutuamente independientes\
condicionada a $y$, entonces:

$P(y/X)$ es proporcional a:

$P(x_0/y)P(x_1/y)...P(x_n/y)P(y)$

En cuanto a la probabilidad marginal $P(y)$, asumimos _a priori_ que\
$P(y=0)=P(y=1)$, lo cual nos permite hacer la clasificación de $y$ usando las productos:

*$P_{ingenua}(y=0/X)$ es proporcional a $P(x_0/y=0)P(x_1/y=0)...P(x_n/y=0)$

*$P_{ingenua}(y=1/X)$ es proporcional a $P(x_0/y=1)P(x_1/y=1)...P(x_n/y=1)$
