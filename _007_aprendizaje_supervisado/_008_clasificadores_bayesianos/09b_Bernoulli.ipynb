**Ejemplo - Bernoulli - Clasificador Bayesiano Ingenuo - Naïve Bayes Classifier**


```
import pandas as pd
url = "https://web.stanford.edu/~hastie/ElemStatLearn/datasets/spam.data"
datos = pd.read_csv(url,sep=' ',header=None)
```

```
import numpy as np
datos = np.asarray(datos)
y = datos[:,-1]
X = datos[:,0:48]
print(y.shape)
print(X.shape)
```

```
import pandas as pd
df = pd.DataFrame(X)
df
```

```
df.describe()
```

```
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = \
    train_test_split(X, y, test_size=0.25, random_state=23)
```

```
from sklearn.naive_bayes import BernoulliNB
# Con el umbral de binarización en cero
BerNB = BernoulliNB(binarize=0.4,fit_prior=True)
BerNB.fit(X_train,y_train)
y_esperado = y_test
y_predicho = BerNB.predict(X_test)
```

```
from sklearn.metrics import accuracy_score, confusion_matrix, recall_score, precision_score
print(accuracy_score(y_esperado,y_predicho))
print(recall_score(y_esperado,y_predicho))
print(precision_score(y_esperado,y_predicho))
print(confusion_matrix(y_esperado,y_predicho))
```

```
from sklearn.naive_bayes import BernoulliNB
# Con el umbral de binarización en cero
BerNB = BernoulliNB(binarize=True,fit_prior=False)
BerNB.fit(X_train,y_train)
y_esperado = y_test
y_predicho = BerNB.predict(X_test)
```

```
from sklearn.metrics import accuracy_score, confusion_matrix, recall_score, precision_score
print(accuracy_score(y_esperado,y_predicho))
print(recall_score(y_esperado,y_predicho))
print(precision_score(y_esperado,y_predicho))
print(confusion_matrix(y_esperado,y_predicho))
```

```
# con el umbral de binarización mayor que cero
BerNB = BernoulliNB(binarize=0.5)
BerNB.fit(X_train,y_train)
y_esperado = y_test
y_predicho = BerNB.predict(X_test)
print(accuracy_score(y_esperado,y_predicho))
```

```
# con el umbral de binarización mayor que cero
BerNB = BernoulliNB(binarize=0.1)
BerNB.fit(X_train,y_train)
y_esperado = y_test
y_predicho = BerNB.predict(X_test)
print(accuracy_score(y_esperado,y_predicho))
```

```
# con el umbral de binarización mayor que cero
BerNB = BernoulliNB(binarize=0.8)
BerNB.fit(X_train,y_train)
y_esperado = y_test
y_predicho = BerNB.predict(X_test)
print(accuracy_score(y_esperado,y_predicho))
```
