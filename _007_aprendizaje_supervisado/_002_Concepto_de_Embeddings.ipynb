Un embedding es una representación matemática de datos (palabras, imágenes, documentos, nodos en un grafo, etc.) en un espacio vectorial de baja o mediana dimensión, de tal forma que capture relaciones y similitudes entre esos datos.

En otras palabras:

- Toma un objeto complejo (ejemplo: una palabra o una imagen)

- Lo convierte en un vector de números reales (ejemplo: un vector en $ℝ^n$)

- Organiza estos vectores de manera que elementos similares estén cerca entre sí y los diferentes estén más alejados.

## Características principales

- Dimensionalidad reducida: en vez de trabajar con datos crudos (texto, píxeles, etc.), se trabaja con un vector mucho más compacto.

- Preservación de relaciones semánticas: en el caso de palabras, "rey – hombre + mujer ≈ reina" es un ejemplo clásico de cómo los embeddings capturan relaciones.

- Generalidad: se pueden aplicar a distintos dominios:

  - NLP (Procesamiento de Lenguaje Natural): Word2Vec, GloVe, BERT producen embeddings de palabras o frases.

  - Visión por computador: redes convolucionales generan embeddings de imágenes.

  - Recomendación: embeddings de usuarios y productos permiten encontrar afinidades.

  - Grafos: Node2Vec, DeepWalk generan embeddings de nodos.

📌 Definición matemática de un embedding

Un embedding es una función de mapeo

$$f:\mathcal{X}→\mathbb{R}^d$$

donde:

$\mathcal{X}$ es el espacio original de los datos (puede ser discreto, continuo o muy complejo). $\mathbb{R}_d$ es un espacio vectorial de dimensión relativamente baja (típicamente $d≪dim(X)$).

La función $f$ busca preservar alguna noción de estructura (distancias, relaciones semánticas, topología, etc.).
