```
# Import the necessary modules and libraries
import numpy as np
from sklearn.tree import DecisionTreeRegressor
from sklearn import tree
import matplotlib.pyplot as plt
import graphviz

# Create a random dataset
rng = np.random.RandomState(1)
X = np.sort(5 * rng.rand(80, 1), axis=0)
y = np.sin(X).ravel()
y[::5] += 3 * (0.5 - rng.rand(16))
```

```
# Fit regression model
regr_1 = DecisionTreeRegressor(criterion="squared_error",
                               #{“mse”, “friedman_mse”, “mae”}, default=”mse”
                               splitter="best",
                               #{“best”, “random”}, default=”best”
                               max_depth=10,
                               #int, default=None
                               min_samples_split=2,
                               #int or float, default=2
                               min_samples_leaf=3,
                               #int or float, default=1
                               min_weight_fraction_leaf=0.0,
                               #float, default=0.0
                               max_features=None,
                               #int, float or {“auto”, “sqrt”, “log2”}, default=None
                               random_state=None,
                               #int or RandomState, default=None
                               max_leaf_nodes=None,
                               #int, default=None
                               min_impurity_decrease=0.0,
                               #float, default=0.0
                               ccp_alpha=0.0
                               #non-negative float, default=0.0
                               )
reggr_1 = regr_1.fit(X, y)

# Predict
X_test = np.arange(0.0, 5.0, 0.01)[:, np.newaxis]
y_1 = regr_1.predict(X_test)

# Plot the results
fig1 = plt.figure()
plt.scatter(X, y, s=20, edgecolor="black",
            c="darkorange", label="data")
plt.plot(X_test, y_1, color="cornflowerblue",
         label="max_depth", linewidth=2)
plt.xlabel("data")
plt.ylabel("target")
plt.title("Decision Tree Regression")
plt.legend()
plt.show()
fig1.savefig("ejemplo02a.png")

#tree.plot_tree(regr_1)
dot_data = tree.export_graphviz(regr_1, out_file=None,
                                filled=True,rounded=True)
graph_1 = graphviz.Source(dot_data)
graph_1.render("arbol_ejemplo02a")
graph_1
```
