```
def heuristica01(jugada):
    jugada1 = jugada[0:3]
    jugada2 = jugada[3:6]
    jugada3 = jugada[6:9]
    jugada4 = jugada[0:9:3]
    jugada5 = jugada[1:9:3]
    jugada6 = jugada[2:9:3]
    jugada7 = jugada[0]+jugada[4]+jugada[8]
    jugada8 = jugada[2]+jugada[4]+jugada[6]
    h1 = 0
    if ((jugada1=="XXX")|(jugada2=="XXX")|(jugada3=="XXX")|(jugada4=="XXX")|(jugada5=="XXX")|(jugada6=="XXX")|(jugada7=="XXX")|(jugada8=="XXX")):
        h1 = 1000
    elif ((jugada1=="OOO")|(jugada2=="OOO")|(jugada3=="OOO")|(jugada4=="OOO")|(jugada5=="OOO")|(jugada6=="OOO")|(jugada7=="OOO")|(jugada8=="OOO")):
        h1 = -1000
    elif (jugada.count('_')==0):
        h1 = 1
    return h1
```

```
def generar_hijos(padre,minmax):
    j = 0
    hijos = []
    for i in padre:
        if (i == "_"):
            t = list(padre)
            if minmax == "max":
                t[j] = "X"
            else:
                t[j] = "O"
            hijo = ''.join(t)
            t.clear()
            hijos.append(hijo)
        j+=1
    return hijos
```

```
def poda_alfa_beta(padre,alfa,beta,nivel,minmax,heuristica):
    print("",end=nivel, file=f)
    if (minmax == "max"):
        print("Nodo max",file=f)
    else:
        print("Nodo min",file=f)
    print("",end=nivel, file=f)
    print(padre[0:3],file=f)
    print("",end=nivel, file=f)
    print(padre[3:6],file=f)
    print("",end=nivel, file=f)
    print(padre[6:9],file=f)    
    print("",end=nivel, file=f)
    print(alfa,beta,file=f)
    alfanew = copy.deepcopy(alfa)
    betanew = copy.deepcopy(beta)
    if (alfanew < heuristica)&(minmax == "max")&(abs(heuristica)>0):
        alfanew = copy.deepcopy(heuristica)
    if (betanew > heuristica)&(minmax == "min")&(abs(heuristica)>0):
        betanew = copy.deepcopy(heuristica)
    if (heuristica == 1000): # nodo hoja
        print("",end=nivel, file=f)      
        print("Gana X",file=f)
    elif (heuristica == 1): # nodo hoja
        print("",end=nivel, file=f)      
        print("Empate",file=f)
    elif (heuristica == -1000): # nodo hoja
        print("",end=nivel, file=f)      
        print("Gana O",file=f)
    else: # se deben generar hijos
        nivelnew= copy.deepcopy(nivel)+"   |"
        print("",end=nivel, file=f)
        print("Generando hijos",file=f)
        hijos = generar_hijos(padre,minmax)
        heuristicas = []
        for hijo in hijos:
            h1 = heuristica01(hijo)
            heuristicas.append(h1)
        hijos_ordenados = list(zip([abs(h1) for h1 in heuristicas],heuristicas,hijos))
        # ajustar si es nodo max o min
        hijos_ordenados.sort(reverse=True)
        podar = 0
        for orden,heur,hijo in hijos_ordenados:
            if (podar == 0):
                if minmax == "max":
                    alfat, betat = poda_alfa_beta(hijo,alfanew,betanew,nivelnew,"min",heur)
                    if (alfanew < betat):
                        alfanew = copy.deepcopy(betat)
                        print("",end=nivel, file=f)
                        print(alfanew,betanew,file=f)
                else:
                    alfat, betat = poda_alfa_beta(hijo,alfanew,betanew,nivelnew,"max",heur)
                    if (alfat < betanew):
                        betanew = copy.deepcopy(alfat)
                        print("",end=nivel, file=f)
                        print(alfanew,betanew,file=f)
                if (alfanew >= betanew):
                    print("",end=nivel, file=f)
                    print("Poda",file=f)
                    podar = 1
    return alfanew,betanew
```
