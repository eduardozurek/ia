# Clustering Jerárquico - Hierarchical Clustering

El clustering jerárquico es una técnica de agrupamiento que busca organizar un conjunto de objetos en una jerarquía de grupos (clusters). A diferencia de métodos como k-means, no requiere especificar de antemano el número de clusters, ya que genera una estructura en forma de árbol llamada dendrograma.

## Tipos principales de clustering jerárquico
Aglomerativo (bottom-up)
* Empieza con cada punto como un cluster independiente.
* En cada paso, se combinan los dos clusters más similares hasta que todos los puntos quedan en un solo cluster.
* Es el enfoque más usado.

Divisivo (top-down)
* Empieza con todos los puntos en un solo cluster.
* Divide iterativamente los clusters en subgrupos hasta llegar a clusters individuales.
* Menos común debido a su mayor costo computacional.

## Métricas de distancia entre clusters

En el proceso de fusión/división, se usan distintas formas de medir la distancia entre clusters:

* Single linkage (mínima distancia): usa la distancia más corta entre dos puntos de distintos clusters.

* Complete linkage (máxima distancia): usa la distancia más larga.

* Average linkage: usa la distancia promedio entre todos los puntos de los clusters.

* Ward’s method: minimiza el incremento en la varianza total dentro de los clusters.

## Visualización: Dendrograma

El resultado se representa en un dendrograma, un árbol que muestra las fusiones (en el caso aglomerativo) o divisiones (en el caso divisivo). Al “cortar” el dendrograma en un nivel determinado, se obtiene un número específico de clusters.

## Ventajas y Desventajas

Ventajas

* No requiere definir K (número de clusters) a priori.

* Produce una jerarquía completa, útil para explorar distintos niveles de agrupación.

* Fácil de interpretar con el dendrograma.

Desventajas

* Costoso en cómputo para grandes conjuntos de datos (O(n²) en tiempo y memoria).

* Sensible al ruido y a outliers.

* No siempre es fácil decidir el “nivel de corte” óptimo en el dendrograma.
