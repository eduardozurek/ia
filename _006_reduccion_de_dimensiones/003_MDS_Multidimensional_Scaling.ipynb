🔹 ¿Qué es MDS?

El Multidimensional Scaling (MDS) busca representar datos en un espacio de baja dimensión (2D o 3D normalmente) preservando lo mejor posible las distancias o similitudes entre los objetos originales.

En lugar de trabajar con variables directamente, parte de una matriz de distancias (o disimilaridades) entre pares de puntos.

🔹 Tipos principales de MDS

1. MDS clásico (también llamado PCoA - Principal Coordinates Analysis):
   - Lineal
   - Usa álgebra lineal para encontrar una representación que preserve distancias euclidianas.

   - Equivalente a PCA pero trabajando con la matriz de distancias en vez de la de covarianza.

2. MDS métrico:
   - No Lineal
   - Busca preservar exactamente las distancias dadas.

   - Minimiza un criterio de error llamado "stress" de Kruskal.

3. MDS no métrico:
   - No Lineal
   - No intenta preservar los valores exactos de las distancias, sino solo el orden de las distancias (quién está más cerca de quién).

   - Útil si los datos originales provienen de medidas ordinales o escalas subjetivas (ej. encuestas de percepción).

🔹 Diferencia con PCA

- PCA: trabaja con variables y busca maximizar la varianza.

- MDS: trabaja con distancias/diferencias entre objetos, sin necesidad de las variables originales.

## 🎯 MDS Clásico (Classical Multidimensional Scaling / PCoA)

El **MDS clásico** es un método de reducción de dimensionalidad que, a partir de una **matriz de distancias** entre objetos, obtiene coordenadas en un espacio de menor dimensión que **preservan lo mejor posible esas distancias**.  
También se conoce como **Principal Coordinates Analysis (PCoA)**.

---

## 📥 Entradas y salidas
- **Entrada:** matriz de distancias $D = [d_{ij}]$ entre $n$ objetos.  
- **Salida:** coordenadas $\hat{X} \in \mathbb{R}^{n \times p}$ con $p \ll n$, donde las distancias euclidianas aproximan $D$.

---

## 🧮 Procedimiento matemático

### 1. Distancias al cuadrado
Construir la matriz de distancias al cuadrado:
$$
(D^{(2)})_{ij} = d_{ij}^2
$$

### 2. Matriz de centrado
Definir:
$$
J = I - \frac{1}{n}\mathbf{1}\mathbf{1}^\top
$$
Donde $1$ es un vector columna de unos

### 3. Matriz de Gram (doble centrado)
Convertir distancias en productos internos:
$$
B = -\tfrac{1}{2} J D^{(2)} J
$$

### 4. Descomposición espectral
Realizar autodescomposición:
$$
B = V \Lambda V^\top
$$

### 5. Coordenadas embebidas
Tomar los $p$ mayores autovalores positivos y sus autovectores:
$$
\hat{X} = V_p \Lambda_p^{1/2}
$$

---

## 📊 Relación con PCA
- Si $D$ proviene de distancias euclidianas de datos centrados, el **MDS clásico** es equivalente a **PCA**.  
- PCA trabaja con la **covarianza**, mientras que MDS clásico trabaja con la **matriz de distancias**.

---

## 📈 Varianza explicada
La proporción de varianza explicada por las primeras $p$ dimensiones es:
$$
\text{Expl}(p) = \frac{\sum_{i=1}^{p} \lambda_i}{\sum_{i:\lambda_i>0} \lambda_i}
$$

---

## ⚠️ Consideraciones
- Si $D$ **no es euclidiana**, aparecen **autovalores negativos** en $B$.  
- Se puede:
  - Ignorar autovalores negativos.  
  - Aplicar correcciones (Cailliez, Lingoes).  

---

## ✅ Resumen
- El **MDS clásico** transforma distancias en coordenadas usando **doble centrado** y **descomposición espectral**.  
- Es equivalente a PCA cuando se usan distancias euclidianas.  
- Sirve como base para métodos no lineales como **Isomap** (que usa distancias geodésicas en lugar de euclidianas).


## MDS Métrico

El MDS métrico es una variante de Multidimensional Scaling que busca encontrar una representación de los datos en un espacio de baja dimensión (generalmente 2D o 3D), de manera que las distancias en ese espacio reproduzcan lo más fielmente posible las distancias originales.

La clave es que en el caso métrico, se asume que las disimilitudes dadas corresponden (o pueden transformarse linealmente) a distancias euclidianas reales.

### ⚖️ Función de costo en MDS (Stress) Métrico y No Métrico

El **Escalamiento Multidimensional (MDS)** mide la calidad de la representación comparando las distancias originales $ d_{ij} $ con las distancias en el espacio reducido $ \hat{d}_{ij} $.  

La función de costo más usada es el **Stress de Kruskal**:

$$
Stress(X) = \sqrt{ \frac{\sum_{i<j} (d_{ij} - \hat{d}_{ij})^2}{\sum_{i<j} d_{ij}^2} }
$$

- Si **Stress ≈ 0**, la representación es casi perfecta.  
- Valores típicos de referencia:  
  - **< 0.05** → Excelente  
  - **0.05 – 0.1** → Muy bueno  
  - **0.1 – 0.2** → Aceptable  
  - **> 0.2** → Pobre  

En resumen, el **Stress** cuantifica qué tan bien las distancias del espacio reducido preservan las distancias originales.


### 📐 Formulación del MDS Métrico

El **MDS Métrico** busca encontrar una representación en baja dimensión que preserve lo mejor posible las **distancias absolutas** entre objetos.  

---

## 1. Entrada
Una matriz de distancias (o disimilitudes) entre pares de objetos:
$$
D = [d_{ij}] \quad \text{para } i,j = 1,\dots,n
$$

---

## 2. Distancias en el espacio embebido
En el espacio reducido (de dimensión $p \ll n$), cada objeto $i$ se representa por un vector $y_i \in \mathbb{R}^p$.  

Las distancias embebidas son:
$$
\hat{d}_{ij} = \| y_i - y_j \|
$$

---

## 3. Función de costo (Stress de Kruskal)
El objetivo es minimizar la diferencia entre distancias originales y distancias embebidas:

$$
Stress(X) = \sqrt{ \frac{\sum_{i<j} (d_{ij} - \hat{d}_{ij})^2}{\sum_{i<j} d_{ij}^2} }
$$

donde:
- $ d_{ij} $ = distancia original entre $i$ y $j$  
- $ \hat{d}_{ij} $ = distancia en el espacio reducido  

---

## 4. Optimización
Se resuelve el problema:

$$
\min_{Y \in \mathbb{R}^{n \times p}} \; Stress(Y)
$$

Normalmente se usa un algoritmo iterativo como **SMACOF** (*Scaling by MAjorizing a COmplicated Function*).  

---

## ✅ Resumen
- El **MDS Métrico** intenta preservar las distancias absolutas.  
- Si el stress es cercano a 0, la representación es casi perfecta.  
- Se usa principalmente cuando las disimilitudes provienen de medidas métricas confiables.


🔹 Algoritmo típico (iterativo)

1. Inicializar posiciones $X$ aleatoriamente o con PCA.

2. Calcular las distancias $d_{ij}$ en el espacio reducido.

3. Minimizar la función de stress con un método iterativo (ej. SMACOF: Scaling by MAjorizing a COmplicated Function).

4. Repetir hasta convergencia (stress estable o mínimo alcanzado).

🔹 Interpretación

- Los ejes en MDS no tienen un significado intrínseco (a diferencia de PCA).

- Lo importante es la relación relativa entre los puntos:

   - puntos cercanos = objetos similares,

   - puntos lejanos = objetos disímiles.

## NMDS - MDS No Métrico

🔹 Definición

El MDS no métrico busca representar objetos en un espacio de baja dimensión (2D o 3D) preservando solo el orden de las distancias o disimilitudes originales, no sus valores exactos.

En otras palabras:

- Si $d_{ij} < d_{kl}$ en los datos originales, entonces en el espacio reducido se exige que $\hat{d}_{ij}<\hat{d}_{kl}$.

- No importa si las magnitudes son distintas, solo que el ranking de distancias sea respetado.

Esto lo hace muy útil cuando las distancias provienen de escalas ordinales o medidas subjetivas, como encuestas, juicios de expertos, similitudes de percepción, etc.

🔹 Fundamento matemático

El NMDS busca una función de transformación monótona $f$ que relacione las distancias originales $d_{ij}$  con las distancias en el espacio reducido $\hat{d}_{ij}$:

$$
\hat{d}_{ij} \approx f(d_{ij})
$$

donde  f es una transformación que preserva el orden (monótona creciente).

La calidad de la representación se mide con el stress de Kruskal (igual que en MDS métrico):

$$
Stress(X) = \sqrt{\frac{\Sigma_{i<j}(f(d_{ij})-\hat{d}_{ij})^2}{\Sigma_{i<j}\hat{d}_{ij}^2}}
$$
	​


🔹 Diferencias con MDS métrico

| Característica           | MDS Métrico ✅                     | MDS No Métrico ✅                   |
| ------------------------ | --------------------------------- | ---------------------------------- |
| Qué preserva             | Distancias absolutas              | Orden de distancias                |
| Tipo de datos            | Distancias métricas (euclidianas) | Distancias ordinales o no métricas |
| Transformación permitida | Lineal o identidad                | Monótona                           |
| Algoritmo típico         | SMACOF u optimización cuadrática  | SMACOF con ordenamientos           |


🔹 Interpretación

- Igual que en el MDS métrico: los ejes no tienen significado absoluto.

- Lo importante es la proximidad relativa: objetos más cercanos en el mapa ≈ más similares.

- NMDS suele ser más robusto con datos ruidosos o disimilitudes no euclidianas.
