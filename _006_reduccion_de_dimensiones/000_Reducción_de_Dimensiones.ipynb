La reducción de dimensiones es un conjunto de técnicas que buscan representar datos de alta dimensión en un espacio de menor número de variables, conservando la mayor cantidad posible de información relevante. Es muy usada en inteligencia artificial, machine learning, minería de datos y análisis estadístico, ya que mejora la eficiencia computacional, facilita la visualización y reduce el ruido en los datos.

📌 Razones para reducir la dimensionalidad:

1. Curse of dimensionality: en espacios de muchas dimensiones, la distancia entre puntos se vuelve poco significativa y los algoritmos pierden efectividad.

2. Ruido: variables irrelevantes pueden afectar el desempeño de los modelos.

3. Costos computacionales: menos dimensiones implican menor tiempo y memoria.

4. Visualización: facilita representar datos en 2D o 3D para interpretación humana.

5. Generalización: ayuda a evitar overfitting.

📊 Métodos de reducción de dimensiones
1. Métodos lineales

   - PCA (Principal Component Analysis / Análisis de Componentes Principales): Proyecta los datos en nuevas variables (componentes principales) que maximizan la varianza.

   - LDA (Linear Discriminant Analysis / Análisis Discriminante Lineal): Es un método SUPERVISADO que busca proyecciones que maximicen la separación entre clases conocidas.

   - MDS (Multidimensional Scaling) Clasico - PCoA: Preserva distancias o similitudes entre puntos.

2. Métodos no lineales

   - MDS:
      - MDS Métrico
      - MDS No Métrico
   - t-SNE (t-distributed Stochastic Neighbor Embedding): Representa relaciones de similitud local en baja dimensión, muy usado en visualización.

   - UMAP (Uniform Manifold Approximation and Projection): Preserva tanto la estructura global como local, más rápido que t-SNE.

   - Isomap: Basado en MDS Clásico, usa distancias geodésicas en un grafo de vecinos para representar la estructura no lineal.

3. Métodos de selección de características. En lugar de transformar variables, seleccionan subconjuntos relevantes:

   - Filtro: selección por correlación, información mutua, chi-cuadrado.

   - Wrapper: evalúa subconjuntos con un modelo predictivo (ej. RFE).

   - Embebidos: selección dentro del propio algoritmo (ej. regularización Lasso).
