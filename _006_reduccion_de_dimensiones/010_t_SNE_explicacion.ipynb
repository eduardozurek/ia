🔹 ¿Qué es t-SNE?

El t-SNE es un algoritmo de reducción de dimensiones diseñado principalmente para visualización en 2D o 3D, que busca preservar la estructura local de los datos: es decir, que los puntos cercanos en alta dimensión también lo estén en la representación reducida.

Fue propuesto por van der Maaten y Hinton (2008).

Imagina que tienes datos en un espacio de 100 dimensiones.
Quieres verlos en un plano 2D, pero sin perder la idea de quién está cerca de quién.

t-SNE construye una distribución de probabilidades de proximidad entre puntos en alta dimensión y luego otra en baja dimensión, tratando de que ambas distribuciones se parezcan lo más posible.

## 🔎 Fundamento Matemático de t-SNE

El algoritmo **t-SNE (t-distributed Stochastic Neighbor Embedding)** busca representar datos de alta dimensión en un espacio de baja dimensión (2D o 3D), **preservando las relaciones de vecindad local**.

---

## 1. Probabilidades de similitud en el espacio original
Dado un conjunto de puntos en alta dimensión $x_1, x_2, \dots, x_n$, se define la probabilidad condicional de que el punto $x_j$ sea vecino de $x_i$:

$$
p_{j|i} = \frac{\exp\!\left(-\frac{\|x_i - x_j\|^2}{2\sigma_i^2}\right)}{\sum_{k \neq i} \exp\!\left(-\frac{\|x_i - x_k\|^2}{2\sigma_i^2}\right)}
$$

donde:
- $\sigma_i$ controla la escala local alrededor de $x_i$.  
- El valor de $\sigma_i$ se ajusta de modo que la **perplexity** de la distribución coincida con el parámetro elegido por el usuario.  

La probabilidad conjunta simétrica se define como:

$$
p_{ij} = \frac{p_{j|i} + p_{i|j}}{2n}
$$

---

## 2. Probabilidades de similitud en el espacio embebido
En el espacio reducido (2D o 3D), cada punto se representa como $y_i$.  
La similitud entre dos puntos se modela con una distribución de **Student-t** con un grado de libertad:

$$
q_{ij} = \frac{(1 + \|y_i - y_j\|^2)^{-1}}{\sum_{k \neq l} (1 + \|y_k - y_l\|^2)^{-1}}
$$

Esta distribución con **colas pesadas** evita que puntos lejanos colapsen demasiado cerca.

---

## 3. Función de costo (Divergencia KL)
El objetivo es que las distribuciones $P = [p_{ij}]$ y $Q = [q_{ij}]$ sean lo más parecidas posible.  
Para ello, t-SNE minimiza la **divergencia de Kullback-Leibler (KL)**:

$$
KL(P \| Q) = \sum_{i \neq j} p_{ij} \log \frac{p_{ij}}{q_{ij}}
$$

---

## 4. Optimización
- Se usa **descenso de gradiente estocástico** para ajustar las coordenadas $y_i$.  
- Inicialmente se colocan los puntos en baja dimensión de forma aleatoria o con PCA.  
- Se actualizan iterativamente hasta que el valor de $KL(P \| Q)$ converge.

---

- t-SNE convierte distancias en **probabilidades de similitud**.  
- Usa **Gaussiana** en alta dimensión y **Student-t** en baja dimensión.
