🔹 ¿Qué es LDA?

Es un método supervisado de reducción de dimensiones y clasificación.
A diferencia de PCA, que ignora las etiquetas de clase, LDA sí utiliza la información de clase para encontrar las direcciones (combinaciones lineales de variables) que maximizan la separación entre clases.

# 📐 Fundamento Matemático de LDA

El **Análisis Discriminante Lineal (LDA)** busca encontrar una proyección lineal que **maximice la separación entre clases**.  

## 1. Objetivo
Queremos un vector de proyección $ w $ que transforme los datos $ x $ en:

$$
y = w^T x
$$

de modo que la separación entre clases en el espacio proyectado sea máxima.  

---

## 2. Dispersión intra-clase
La **matriz de dispersión intra-clase** mide qué tan compactos son los puntos de cada clase:

$$
S_W = \sum_{c=1}^{C} \sum_{x \in X_c} (x - \mu_c)(x - \mu_c)^T
$$

donde:  
- $ X_c $ = conjunto de muestras de la clase $ c $  
- $ \mu_c $ = media de la clase $ c $  

---

## 3. Dispersión entre clases
La **matriz de dispersión entre clases** mide qué tan alejadas están las medias de las clases:

$$
S_B = \sum_{c=1}^{C} N_c (\mu_c - \mu)(\mu_c - \mu)^T
$$

donde:  
- $ N_c $ = número de muestras en la clase $ c $  
- $ \mu $ = media global  

---

## 4. Función de optimización
El objetivo de LDA es maximizar la razón de **dispersión entre clases** sobre la **dispersión intra-clase**:

$$
J(w) = \frac{w^T S_B w}{w^T S_W w}
$$

---

## 5. Solución
El problema se resuelve resolviendo el sistema de autovalores:

$$
S_W^{-1} S_B w = \lambda w
$$

Los vectores propios $ w $ correspondientes a los mayores autovalores $\lambda$ definen las **direcciones discriminantes**.  

---

👉 En conclusión: LDA proyecta los datos en un subespacio de dimensión reducida (máximo $ C-1 $, con $ C $ clases), preservando la máxima separabilidad entre clases.


🔹 Diferencias clave con PCA

- PCA: no supervisado, maximiza la varianza total.

- LDA: supervisado, maximiza la separación entre clases.

- Dimensiones máximas en LDA: si tienes $c$ clases, LDA puede reducir a lo sumo a $c -1$ dimensiones.
